### 1. Definición de las funciones Piso y Techo

*   **Función Piso (\\(\lfloor x \rfloor\\) o *floor*):** Para cualquier número real \\(x\\), representa el mayor entero que es menor o igual que \\(x\\). Por ejemplo, \\(\lfloor 2.9 \rfloor = 2\\), \\(\lfloor 2.1 \rfloor = 2\\) y \\(\lfloor -3.8 \rfloor = -4\\).
*   **Función Techo (\\(\lceil x \rceil\\) o *ceiling*):** Para cualquier número real \\(x\\), representa el menor entero que es mayor o igual que \\(x\\). Por ejemplo, \\(\lceil 6.1 \rceil = 7\\), \\(\lceil 6.2 \rceil = 7\\) y \\(\lceil -3.8 \rfloor = -3\\).

#### Propiedades matemáticas relevantes:
*   **Acotación:** Ambas funciones acotan al número real \\(x\\) según la relación: \\(x - 1 < \lfloor x \rfloor \le x \le \lceil x \rceil < x + 1\\).
*   **Monotonicidad:** Ambas son funciones monótonamente crecientes, lo que significa que si \\(m \le n\\), entonces sus respectivos pisos y techos conservan esa misma relación de orden.
*   **División por la mitad:** Para cualquier entero \\(n\\), se cumple que \\(\lceil n/2 \rceil + \lfloor n/2 \rfloor = n\\).

### 2. Usos de Piso y Techo en Computación

Las fuentes de tu notebook destacan que estas funciones se aplican en diversos escenarios prácticos de la ingeniería de software y el análisis de algoritmos:

    \\[T(n) = aT(\lfloor n/b \rfloor) + f(n) \quad \text{o} \quad T(n) = aT(\lceil n/b \rceil) + f(n)\\].
    *(Nota: Aunque son esenciales para la correctitud matemática, los textos aclaran que a menudo se omiten en el análisis simplificado de complejidad asintótica porque no alteran el comportamiento del crecimiento del algoritmo).*

### 3. Crecimiento de la Función Exponencial vs. Polinomial

El motivo por el cual una función exponencial positiva creará siempre una curva de crecimiento mucho más empinada y rápida que cualquier polinomio se debe a un principio fundamental de **límites asintóticos**.

Para cualquier constante real polinomial \\(b\\) (que represente el grado del polinomio, por ejemplo, \\(n^2\\), \\(n^3\\), \\(n^{100}\\)) y cualquier constante exponencial con base \\(a > 1\\) (como \\(2^n\\) o \\(3^n\\)), se establece que **el límite de su cociente cuando \\(n\\) tiende al infinito es igual a cero**:

\\[\lim_{n \to \infty} \frac{n^b}{a^n} = 0\\]

#### ¿Por qué ocurre esto y qué significa?
*   **Significado del límite:** Conforme el tamaño de la entrada \\(n\\) se hace sumamente grande, el valor de la exponencial \\(a^n\\) (en el denominador) se vuelve infinitamente más grande y dominante en comparación con el valor de la potencia polinomial \\(n^b\\) (en el numerador).
*   **Relación de "o pequeña" (*o-notation*):** A partir de este límite, matemáticamente se concluye que:
    \\[n^b = o(a^n)\\].
    Esto formaliza que la tasa de crecimiento de cualquier polinomio está estrictamente por debajo de la tasa de crecimiento de cualquier función exponencial positiva.
*   **Impacto práctico en el software:** Un algoritmo con complejidad polinomial, como \\(O(n^2)\\) o \\(O(n^3)\\), se considera "tolerable" y puede ejecutarse en un tiempo razonable para datos grandes. En cambio, un algoritmo con complejidad exponencial, como \\(O(2^n)\\), es considerado **intratable**; para un valor moderado de \\(n = 100\\), el número de operaciones requeridas por un algoritmo exponencial es tan astronómicamente alto que ninguna computadora existente podría terminar de procesarlo antes de que transcurran millones de años.

**RESPUESTA:** ¿Qué es un algoritmo y cuáles son sus características clave?

Un **algoritmo** se define intuitivamente como una **serie o conjunto finito de pasos ordenados para resolver un problema**. De manera más formal, representa un método sistemático que actúa como un "conjunto de reglas operacionales inherentes a un cómputo", el cual es capaz de ser ejecutado mecánicamente por una entidad (como una computadora) para transformar un conjunto de datos de entrada en una salida correcta.

Para que un procedimiento o método sea considerado formalmente un algoritmo, debe cumplir rigurosamente con las siguientes **propiedades esenciales**:
1. **Entrada y Salida:** Debe tener claramente definido qué datos necesita para iniciar (cero o más entradas) y qué resultados o productos debe generar (una o más salidas).
2. **Precisión (No ambiguo):** Cada instrucción debe ser explícita, clara y estar libre de ambigüedades. El orden exacto de los pasos a seguir debe estar perfectamente determinado para que la máquina siempre sepa qué comando ejecutar a continuación.
3. **Determinismo (Unicidad):** Si el algoritmo se ejecuta múltiples veces utilizando exactamente el mismo conjunto de datos de entrada, debe producir siempre los mismos datos de salida.
4. **Finitud:** El algoritmo debe terminar obligatoriamente tras concluir un número finito de pasos. Si un procedimiento es preciso y determinista pero no tiene fin, se le denomina simplemente un **"método computacional"**. Además, cada paso individual debe poder completarse en un tiempo finito.
5. **Efectividad:** Cada paso o acción debe ser lo suficientemente básico como para poder llevarse a cabo de forma física (por ejemplo, con lápiz y papel) en un tiempo finito, logrando el efecto deseado.
6. **Generalidad:** Debe ser lo suficientemente general para contemplar y resolver de forma correcta todos los casos posibles de datos de entrada del problema.

### El "Problema de Paro" (Halting Problem) de Alan Turing

Formulado originalmente por el matemático británico **Alan Turing en 1936**, el **Problema de Paro** es un hito fundamental en la teoría de la computabilidad. Este problema de decisión plantea la siguiente pregunta fundamental:

> *Dado un programa de computadora cualquiera y un conjunto de datos de entrada arbitrarios, ¿es posible determinar si el programa terminará su ejecución en algún momento (es decir, se detendrá) o si continuará trabajando indefinidamente en un bucle infinito?*

Turing demostró matemáticamente que **no puede existir ningún algoritmo general capaz de resolver correctamente todas las instancias de este problema**. Por esta razón, el Problema de Paro está catalogado como un problema **indecidible (undecidable)** o no computable.

La demostración teórica se realiza mediante una contradicción matemática (reducción al absurdo). Turing probó que si existiera un algoritmo capaz de predecir el paro, se podría construir un programa contradictorio que hiciera exactamente lo opuesto a lo predicho, rompiendo la lógica del sistema. Además, en términos prácticos, propuestas intuitivas como "correr el programa para ver si se detiene" no resuelven el problema: si el programa que se está analizando entra en un bucle infinito, el evaluador se quedaría esperando para siempre sin poder determinar de forma definitiva si el programa parará o no en el futuro.

### Análisis A Priori vs. Prueba A Posteriori

La evaluación de la **eficiencia de un algoritmo** (en términos de cuánta memoria y tiempo de ejecución consume para resolver un problema) consta de dos fases o aproximaciones metodológicas muy distintas:

| Criterio | Análisis A Priori (Teórico) | Prueba A Posteriori (Empírico / Experimental) |
| :--- | :--- | :--- |
| **¿Cuándo se realiza?** | En la **etapa de diseño** del algoritmo, antes de escribir código real. | **Después de la implementación** física del algoritmo en un lenguaje de programación. |
| **¿En qué consiste?** | Se estudia matemáticamente una representación de alto nivel (como pseudocódigo) para deducir una función límite (como \\(T(n)\\)) que describa el comportamiento del algoritmo según el tamaño de la entrada \\(n\\). | Se programa el algoritmo, se ejecuta en una computadora física con datos de prueba seleccionados, y se recogen estadísticas de uso real de tiempo (reloj en mano) y espacio en disco/RAM. |
| **Principales Ventajas** | *   Permite predecir costos y descartar soluciones malas sin el esfuerzo de codificarlas.<br>*   Es **independiente del hardware, software, compilador o lenguaje** de programación.<br>*   Considera de manera matemática todas las posibles entradas de datos. | *   Es aplicable a absolutamente cualquier algoritmo, sin importar qué tan complejo resulte su análisis matemático.<br>*   Permite evaluar el comportamiento del software en condiciones de arquitectura reales. |
| **Principales Limitaciones** | *   Su aplicabilidad matemática detallada puede resultar sumamente difícil, especialmente al investigar el comportamiento para casos promedio. | *   Los experimentos se limitan a un **conjunto acotado de entradas de prueba**.<br>*   Los resultados **dependen enteramente de la computadora** utilizada (velocidad del CPU, lenguaje de programación, compilador, etc.). |

Mientras que el análisis teórico (A Priori) nos da una cota universal que define la escalabilidad del algoritmo en el infinito, la prueba empírica (A Posteriori) nos ayuda a verificar el rendimiento bajo entornos y limitaciones físicas de producción.

- *Explica en qué consisten el peor caso, mejor caso y caso promedio al momento de evaluar el comportamiento de un algoritmo ante distintas entradas.*  
**RESPUESTA:** Cuando se analiza la eficiencia de un algoritmo, el tiempo de ejecución no solo depende del tamaño de la entrada (\(n\)), sino también de la **disposición o naturaleza de los datos**. Por ello, se estudian tres escenarios fundamentales para describir este comportamiento dentro de un rango de valores acotado por \(T_{\min}(n) \le T(n) \le T_{\max}(n)\):

---

### 1. Peor Caso (*Worst Case*)
*   **Definición:** Es la situación en la cual el algoritmo realiza la **mayor cantidad de operaciones elementales o consume el máximo tiempo de ejecución** de entre todas las entradas posibles de tamaño \(n\). Define la cota máxima \(T_{\max}(n)\) de consumo de recursos.
*   **Importancia:** Proporciona una **garantía de seguridad (cota superior)**; es decir, asegura matemáticamente que el algoritmo nunca tardará más que este límite, sin importar qué tan desfavorables sean los datos de entrada. Es el tipo de análisis más utilizado y conservador en computación.
*   **Ejemplo:** En una **búsqueda secuencial** en un arreglo de tamaño \(n\), el peor caso ocurre cuando el elemento que se busca está en la última posición o no existe en absoluto, lo que obliga al algoritmo a recorrer el arreglo completo y realizar las \(n\) comparaciones.

### 2. Mejor Caso (*Best Case*)
*   **Definición:** Es el escenario en el que el algoritmo realiza la **menor cantidad de operaciones o toma el menor tiempo posible** para resolver el problema con una entrada de tamaño \(n\). Define la cota mínima \(T_{\min}(n)\).
*   **Importancia:** Aunque por lo general es poco probable que se presenten las condiciones ideales en la práctica, analizar el mejor caso es útil para identificar el comportamiento del algoritmo ante entradas "casi resueltas" (como el ordenamiento por inserción con datos casi ordenados). Además, si el mejor caso de un algoritmo ya es ineficiente, sirve para descartar su uso de inmediato.
*   **Ejemplo:** En la **búsqueda secuencial**, el mejor caso se da cuando el valor buscado está en la primera posición del arreglo, permitiendo que el algoritmo termine tras realizar una única comparación (\(O(1)\)).

### 3. Caso Promedio (*Average Case*)
*   **Definición:** Describe el **comportamiento esperado o típico del algoritmo** frente a entradas "aleatorias" o comunes del mundo real.
*   **Importancia:** Es considerado la **medida más realista** del desempeño de un programa en entornos de producción.
*   **Complejidad matemática:** Es significativamente más difícil de calcular que los otros dos escenarios. No se obtiene simplemente promediando el mejor y el peor caso. Para calcularlo, se debe definir o asumir una **distribución de probabilidad** sobre todos los posibles datos de entrada de tamaño \(n\) (por ejemplo, asumiendo que todas las entradas son igualmente probables) y calcular el valor esperado de operaciones.
*   **Ejemplo:** En la **búsqueda secuencial**, bajo el supuesto de que el elemento buscado tiene la misma probabilidad de estar en cualquier posición del arreglo, en promedio se tendrá que examinar aproximadamente la mitad de los datos (\(n/2\) comparaciones), lo cual mantiene un comportamiento de crecimiento lineal.

## 4. Sección Práctica 2: Iteración vs. Recursión
Un algoritmo recursivo debe llamarse a sí mismo resolviendo un problema menor e incluir siempre un "caso base" para evitar ciclos infinitos.

### [CÓDIGO 2.1] Sucesión de Fibonacci Recursiva
```python
# Fi = Fi-1 + Fi-2 para i >= 2; F0 = 0, F1 = 1
def fibonacci_recursivo(n):
    if n == 0: return 0
    elif n == 1: return 1
    else: return fibonacci_recursivo(n-1) + fibonacci_recursivo(n-2)

````python
def fibonacci_recursivo(n):
    if n < 0:
        raise ValueError("n debe ser un entero no negativo")
    if n < 2:
        return n
    return fibonacci_recursivo(n - 1) + fibonacci_recursivo(n - 2)

print("Fibonacci(0):", fibonacci_recursivo(0))
print("Fibonacci(1):", fibonacci_recursivo(1))
print("Fibonacci(10):", fibonacci_recursivo(10))
````


### [CÓDIGO 2.2] Factorial (Iterativo)

1. Implementa una versión recursiva para el cálculo del factorial ($n! = n \cdot (n-1)!$) y una versión iterativa para Fibonacci.

````python
def factorial_recursivo(n):
    if n < 0:
        raise ValueError("n debe ser un entero no negativo")
    if n <= 1:
        return 1
    return n * factorial_recursivo(n - 1)

def fibonacci_iterativo(n):
    if n < 0:
        raise ValueError("n debe ser un entero no negativo")
    actual, siguiente = 0, 1
    for _ in range(n):
        actual, siguiente = siguiente, actual + siguiente
    return actual

for n in (0, 1, 5):
    print(f"Factorial recursivo de {n}: {factorial_recursivo(n)}")
for n in (0, 1, 10):
    print(f"Fibonacci iterativo de {n}: {fibonacci_iterativo(n)}")
````

2. Utiliza la librería `time` o la función mágica `%timeit` para comparar el tiempo consumido (A posteriori) entre el Fibonacci recursivo y el iterativo para `n=35`.

````python
import time

def medir_tiempo(funcion, n, repeticiones):
    mediciones = []
    for _ in range(3):
        inicio = time.perf_counter()
        for _ in range(repeticiones):
            funcion(n)
        mediciones.append((time.perf_counter() - inicio) / repeticiones)
    return min(mediciones)

n = 35
resultado_recursivo = fibonacci_recursivo(n)
resultado_iterativo = fibonacci_iterativo(n)
assert resultado_recursivo == resultado_iterativo == 9227465
tiempo_recursivo = medir_tiempo(fibonacci_recursivo, n, 1)
tiempo_iterativo = medir_tiempo(fibonacci_iterativo, n, 10000)
print(f"n = {n}")
print(f"Resultado recursivo: {resultado_recursivo}")
print(f"Resultado iterativo: {resultado_iterativo}")
print(f"Recursivo: {tiempo_recursivo:.9f} segundos por ejecución")
print(f"Iterativo: {tiempo_iterativo:.9f} segundos por ejecución")
print(f"Relación recursivo/iterativo: {tiempo_recursivo / tiempo_iterativo:.2f}")
print("Se reporta el menor tiempo por ejecución de tres tandas.")
````
